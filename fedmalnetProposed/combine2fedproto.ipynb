```python

# %%
import pandas as pd
import numpy as np
 
 
def report(df, name):
    """Cetak ringkasan singkat sebuah dataframe."""
    print(f"\n================ {name} ================")
    print("Ukuran           :", df.shape)
    print("Missing value    :", df.isna().sum().sum())
    print("Tipe data        :\n", df.dtypes.value_counts())
 
 
def split_X_y(df, label_col):
    X = df.drop(columns=[label_col])
    y = df[label_col]
    print(f"Ukuran Fitur (X): {X.shape} | Ukuran Label (y): {y.shape}")
    return X, y
 
 
def map_labels(series, mapping, name):
    """Map label ke 0/1 dan validasi tidak ada nilai yang gagal ter-mapping."""
    mapped = series.map(mapping)
    n_unmapped = mapped.isna().sum()
    if n_unmapped > 0:
        unknown_vals = series[mapped.isna()].unique()
        raise ValueError(
            f"[{name}] Ada {n_unmapped} nilai label yang tidak dikenali: {unknown_vals}. "
            f"Cek mapping: {mapping}"
        )
    return mapped.astype(int)
 
```

```python

# %%
data_malgenome = pd.read_csv('../dataset/malgenome.csv')
report(data_malgenome, "MALGENOME - AWAL")
 
data_malgenome = data_malgenome.dropna().reset_index(drop=True)
data_malgenome["class"] = map_labels(
    data_malgenome["class"], {"B": 0, "S": 1}, "Malgenome"
)
 
print("\nDistribusi kelas:")
print(data_malgenome["class"].value_counts())
 
X_malgenome, y_malgenome = split_X_y(data_malgenome, "class")
```

```python

# %%
DATA_PATH = "../dataset/drebin_dataset.csv"
LABEL_COLUMN = "class"
 
drebin = pd.read_csv(DATA_PATH)
report(drebin, "DREBIN - AWAL (mentah)")
 
# Hapus kolom index yang tidak diperlukan
if "Unnamed: 0" in drebin.columns:
    drebin = drebin.drop(columns=["Unnamed: 0"])
 
if LABEL_COLUMN not in drebin.columns:
    raise KeyError(
        f"Kolom '{LABEL_COLUMN}' tidak ditemukan. Kolom tersedia: {list(drebin.columns)}"
    )
 
# Konversi label dulu, sebelum kolom fitur dikonversi ke numerik
drebin[LABEL_COLUMN] = map_labels(drebin[LABEL_COLUMN], {"B": 0, "S": 1}, "Drebin")
 
# Konversi seluruh kolom fitur ke numerik.
# Ini juga otomatis mengubah nilai non-numerik semacam "?" jadi NaN (errors="coerce"),
# jadi tidak perlu penanganan khusus per kolom.
feature_columns = [c for c in drebin.columns if c != LABEL_COLUMN]
drebin[feature_columns] = drebin[feature_columns].apply(pd.to_numeric, errors="coerce")
 
report(drebin, "DREBIN - SETELAH KONVERSI NUMERIK (sebelum dropna)")
 
# Baru sekarang cek & hapus missing value, setelah semua kolom benar-benar numerik
drebin = drebin.dropna().reset_index(drop=True)
report(drebin, "DREBIN - SETELAH CLEANING")
 
print("\nDistribusi kelas:")
print(drebin[LABEL_COLUMN].value_counts())
print(drebin[LABEL_COLUMN].value_counts(normalize=True).mul(100).round(2))
 
X_drebin, y_drebin = split_X_y(drebin, LABEL_COLUMN)
y_drebin = y_drebin.astype(int)
 
data_drebin = drebin.copy()
```

```python

# %% [markdown]
# ## Dataset 3 (Tuandroid)
 
# %%
data_tuandroid = pd.read_csv('../dataset/TUANDROMD 2.csv')
report(data_tuandroid, "TUANDROID - AWAL")
 
data_tuandroid = data_tuandroid.dropna().reset_index(drop=True)
data_tuandroid["Label"] = map_labels(
    data_tuandroid["Label"], {"goodware": 0, "malware": 1}, "Tuandroid"
)
 
print("\nDistribusi kelas:")
print(data_tuandroid["Label"].value_counts())
 
X_tuandroid, y_tuandroid = split_X_y(data_tuandroid, "Label")
```

```python

# %% [markdown]
# ## Dataset 4 (Kronodroid)
 
# %%
kronodroid = pd.read_csv('../dataset/kronodroid.csv')
report(kronodroid, "KRONODROID - AWAL")
 
kronodroid = kronodroid.dropna().reset_index(drop=True)
 
print("\nDistribusi kelas:")
print(kronodroid["Malware"].value_counts())
 
X_kronodroid, y_kronodroid = split_X_y(kronodroid, "Malware")
 
 
# %% [markdown]
# ## Ringkasan Semua Dataset
 
# %%
summary = pd.DataFrame({
    "dataset": ["Malgenome", "Drebin", "Tuandroid", "Kronodroid"],
    "n_samples": [X_malgenome.shape[0], X_drebin.shape[0], X_tuandroid.shape[0], X_kronodroid.shape[0]],
    "n_features": [X_malgenome.shape[1], X_drebin.shape[1], X_tuandroid.shape[1], X_kronodroid.shape[1]],
})
print(summary)
```
