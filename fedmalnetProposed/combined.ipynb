## Dataset 2 (Drebin)

```python
# %%
import pandas as pd
import numpy as np
 
# %% [markdown]
# ## 1. Load Dataset
 
# %%
DATA_PATH = "../dataset/drebin_dataset.csv"
LABEL_COLUMN = "class"  # ganti ke "Malware" kalau memang itu nama kolom labelnya

drebin = pd.read_csv(DATA_PATH)
print("Ukuran awal:", drebin.shape)
drebin.head()
 
# %% [markdown]
# ## 2. Eksplorasi Awal (opsional, sebelum dibersihkan)
 
# %%
print("Jumlah missing value per kolom:\n", drebin.isna().sum())
print("\nTipe data per kolom:\n", drebin.dtypes.value_counts())
print("\nKolom bertipe object:\n", drebin.select_dtypes(include="object").columns.tolist())
```

```python
# %%
# =========================================================
# 2. INFORMASI DATASET AWAL
# =========================================================
print("\n================ DATASET AWAL ================")

print("Ukuran data :", drebin.shape)

print("\nTipe data:")
print(drebin.dtypes.value_counts())

print("\nMissing value:")
print(drebin.isna().sum())

print("\nDistribusi label:")
if "class" in drebin.columns:
    print(drebin["class"].value_counts())


# %%
# =========================================================
# 3. PEMBERSIHAN KOLOM
# =========================================================

# Hapus kolom index yang tidak diperlukan
if "Unnamed: 0" in drebin.columns:
    drebin = drebin.drop(columns=["Unnamed: 0"])

print("\nKolom setelah pembersihan:")
print(drebin.columns.tolist())


# %%
# =========================================================
# 4. PEMBERSIHAN FITUR DREBIN
# =========================================================

# Kolom yang memiliki nilai "?"
SPECIAL_COLUMN = "TelephonyManager.getSimCountryIso"

if SPECIAL_COLUMN in drebin.columns:

    # Ubah "?" menjadi NaN
    drebin[SPECIAL_COLUMN] = drebin[SPECIAL_COLUMN].replace(
        "?", np.nan
    )

    # Konversi menjadi numerik
    drebin[SPECIAL_COLUMN] = pd.to_numeric(
        drebin[SPECIAL_COLUMN],
        errors="coerce"
    )


# %%
# =========================================================
# 5. KONVERSI FITUR KE NUMERIK
# =========================================================

# Pisahkan label terlebih dahulu
if "class" not in drebin.columns:
    raise KeyError(
        "Kolom 'class' tidak ditemukan. "
        f"Kolom yang tersedia: {list(drebin.columns)}"
    )

# Konversi label
drebin["class"] = drebin["class"].replace({
    "B": 0,
    "S": 1
})

# Pastikan seluruh fitur numerik
feature_columns = [
    col for col in drebin.columns
    if col != "class"
]

drebin[feature_columns] = drebin[feature_columns].apply(
    pd.to_numeric,
    errors="coerce"
)


# %%
# =========================================================
# 6. MENANGANI MISSING VALUE
# =========================================================

print("\n================ MISSING VALUE ================")

print("Missing value sebelum penghapusan:")
print(drebin.isna().sum().sum())

# Hapus baris yang masih memiliki missing value
drebin = drebin.dropna().reset_index(drop=True)

print("\nMissing value setelah penghapusan:")
print(drebin.isna().sum().sum())

print("\nUkuran dataset setelah cleaning:")
print(drebin.shape)


# %%
# =========================================================
# 7. DISTRIBUSI KELAS
# =========================================================

print("\n================ DISTRIBUSI KELAS ================")

class_distribution = drebin["class"].value_counts()

print(class_distribution)

print("\nPersentase kelas:")
print(
    drebin["class"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# %%
# =========================================================
# 8. MEMISAHKAN FITUR DAN LABEL
# =========================================================

X = drebin.drop(columns=["class"])
y = drebin["class"].astype(int)

print("\n================ DATA X DAN y ================")

print("Ukuran fitur (X)  :", X.shape)
print("Ukuran label (y)  :", y.shape)

print("\nTipe data X:")
print(X.dtypes.value_counts())

print("\nDistribusi y:")
print(y.value_counts())


# %%
# =========================================================
# 9. VALIDASI DATASET
# =========================================================

print("\n================ VALIDASI ================")

print("Jumlah samples :", X.shape[0])
print("Jumlah fitur   :", X.shape[1])
print("Jumlah kelas   :", y.nunique())

print("\nMissing value X :", X.isna().sum().sum())
print("Missing value y :", y.isna().sum())

print("\nLabel unik :", sorted(y.unique()))


# %%
# =========================================================
# 10. DATASET FINAL
# =========================================================

data_drebin = drebin.copy()

X_drebin = X.copy()
y_drebin = y.copy()

print("\n================ DREBIN-215 FINAL ================")

print("Data :", data_drebin.shape)
print("X    :", X_drebin.shape)
print("y    :", y_drebin.shape)
```
